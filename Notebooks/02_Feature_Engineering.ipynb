# Feature Engineering
The objective of this notebook is to create new features from the cleaned dataset that will improve business analysis and support machine learning models.


```python
import pandas as pd
import numpy as np
```

```python
clean_df = pd.read_csv("Retail_Cleaned.csv")
```

```python
clean_df["InvoiceDate"] = pd.to_datetime(clean_df["InvoiceDate"])
```

```python
clean_df.info()
```

## Feature 1: TotalAmount

TotalAmount represents the revenue generated from each transaction.

Formula:
TotalAmount = Quantity × Price

```python
clean_df["TotalAmount"] = clean_df["Quantity"] * clean_df["Price"]
```

```python
clean_df[["Quantity","Price","TotalAmount"]].head()
```

## Feature 2: Year

```python
clean_df["Year"] = clean_df["InvoiceDate"].dt.year
```

```python
clean_df[["InvoiceDate","Year"]].head()
```

```python
clean_df["Month"] = clean_df["InvoiceDate"].dt.month
```

```python
clean_df[["InvoiceDate","Month"]].head()
```

```python
clean_df["MonthName"] = clean_df["InvoiceDate"].dt.month_name()
```

```python
clean_df[["Month","MonthName"]].head()
```

```python
clean_df["Day"] = clean_df["InvoiceDate"].dt.day
```

```python
clean_df[["InvoiceDate","Day"]].head()
```

```python
clean_df["Weekday"] = clean_df["InvoiceDate"].dt.day_name()
```

```python
clean_df[["InvoiceDate","Weekday"]].head()
```

```python
clean_df["Hour"] = clean_df["InvoiceDate"].dt.hour
```

```python
clean_df[["InvoiceDate","Hour"]].head()
```

```python
clean_df["Weekend"] = clean_df["Weekday"].isin(["Saturday","Sunday"])
```

```python
clean_df["Weekend"] = clean_df["Weekend"].astype(int)
```

```python
clean_df[["Weekday","Weekend"]].head()
```

```python
clean_df["Quarter"] = clean_df["InvoiceDate"].dt.quarter
```

```python
clean_df[["InvoiceDate","Quarter"]].head()
```

```python
def time_period(hour):
    if hour < 12:
        return "Morning"
    elif hour < 17:
        return "Afternoon"
    else:
        return "Evening"

clean_df["TimeOfDay"] = clean_df["Hour"].apply(time_period)
```

```python
clean_df[["Hour","TimeOfDay"]].head()
```

```python
clean_df.head()
```

```python
clean_df.columns
```

```python
clean_df.info()
```

```python
clean_df[
    ["TotalAmount","Year","Month","MonthName","Day",
     "Weekday","Hour","Weekend","Quarter","TimeOfDay"]
].head()
```
