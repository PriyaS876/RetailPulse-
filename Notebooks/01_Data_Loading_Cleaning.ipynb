# RetailPulse - Data Loading & Cleaning

## Objective
The objective of this notebook is to:
- Load the Online Retail II dataset.
- Understand the dataset structure.
- Identify missing values and duplicates.
- Clean the dataset.
- Save the cleaned dataset for further analysis.

```python
import pandas as pd
import numpy as np
```

```python
file_path = "../Data/raw/online_retail_II.xlsx"

df = pd.read_excel(file_path)

df.head()
```

```python
# Create a copy of the original dataset
clean_df = df.copy()
```

```python
# Remove duplicate records
clean_df = clean_df.drop_duplicates()

print("Duplicate rows removed successfully.")
print("Remaining rows:", clean_df.shape[0])
```

```python
clean_df.duplicated().sum()
```

```python
clean_df.dtypes
```

```python
clean_df.nunique()
```

```python
# Convert InvoiceDate to datetime
clean_df["InvoiceDate"] = pd.to_datetime(clean_df["InvoiceDate"])

# Verify datatype
clean_df["InvoiceDate"].dtype
```

```python
# Count cancelled invoices
clean_df[clean_df["Invoice"].astype(str).str.startswith("C")].shape
```

```python
# Remove cancelled invoices
clean_df = clean_df[~clean_df["Invoice"].astype(str).str.startswith("C")]

print("Cancelled invoices removed.")
print("Remaining rows:", clean_df.shape)
```

```python
# Check records with negative quantity
clean_df[clean_df["Quantity"] <= 0].shape
```

```python
# Keep only positive quantities
clean_df = clean_df[clean_df["Quantity"] > 0]

print("Negative quantities removed.")
print(clean_df.shape)
```

```python
# Check invalid prices
clean_df[clean_df["Price"] <= 0].shape
```

```python
# Keep only positive prices
clean_df = clean_df[clean_df["Price"] > 0]

print("Invalid prices removed.")
print(clean_df.shape)
```

```python
# Remove rows where Customer ID is missing
clean_df = clean_df.dropna(subset=["Customer ID"])

print("Missing Customer ID removed.")
print(clean_df.shape)
```

```python
clean_df["Description"].isnull().sum()
```

```python
clean_df.info()
```

```python
clean_df.isnull().sum()
```

```python
clean_df.to_csv("../Data/processed/clean_online_retail.csv", index=False)

print("Cleaned dataset saved successfully.")
```
